# Завдання 2
Розділіть дані на залежну y(`expenses`) та незалежну X ознаки. Розділіть дані на тренувальну та тестову частини

```python
X = df.drop('expenses', axis=1)
y = df['expenses']

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    train_size=0.8,
                                                    random_state=42)
```

```python
X_test, y_test
```

# Завдання 3
Натренуйте LinearRegression та виведіть такі метрики: MAE, RMSE, R2

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

model = LinearRegression()
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

y_pred
```

```python
# Metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("mae:",mae)
print("rmse:",rmse)
print("r2:",r2)
```

# Завдання 5 (оптимально)
Перевірте як видалення викидів вплине на результат. Розгляньте один з варіантів:
* видалення викидів по стовпчику `expenses`
* видалення викидів по всіх стовпчиків

```python
numeric_cols = df.select_dtypes(include=['number'])

# Q1, Q3 - квартилів і IQR for numeric
Q1 = numeric_cols.quantile(0.25)
Q3 = numeric_cols.quantile(0.75)
IQR = Q3 - Q1

# Lower and upper bound for outliners
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Outliners in nmeric columns
outliers = (numeric_cols < lower_bound) | (numeric_cols > upper_bound)

# NUmber of outliners per column
print(outliers.sum())

# Rows with outliners
df_outliers = df[outliers.any(axis=1)]
df_outliers
```

```python
df_cleaned = df[~outliers.any(axis=1)]

df_cleaned.shape
```
