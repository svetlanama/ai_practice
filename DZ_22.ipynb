```python
from langchain.llms import HuggingFaceHub

llm = HuggingFaceHub(
    repo_id="mistralai/Mistral-7B-Instruct-v0.2",
    model_kwargs={"temperature": 0.7, "max_length": 200}
)

```

```python

def generate_pet_names(prompt, breed, color, activity_type, num_names):

    generated_names = llm(prompt)

    names_list = [name.strip() for name in generated_names.split('\n') if name.strip()]

    cleaned_names = []
    for name in names_list:
        if not name.lower().startswith("generate") and not name.lower().startswith("the names"):
            cleaned_names.append(name)

    return cleaned_names


```

```python
num_names = int(input("How many names would you like to generate? "))

# Define breed, color, and activity_type
breed = 'Maynkun'
color = 'red'
activity_type = 'happy'

prompt = (f"Come up with {num_names} original and playful names for a {color} {breed} pet who loves {activity_type}. "
          "The names should be fun, easy to remember, and match the pet's character and looks. "
          "List each name on a separate line, without additional comments or details.")

names = generate_pet_names(prompt=prompt, breed=breed, color=color, activity_type=activity_type, num_names=num_names)
print("\nGenerated pet Names: ")
for name in names:
    print(name)


while True:
    try:
        selected_name_index = int(input("\nSelect the number of the name you'd like to choose: ")) - 1
        if selected_name_index < 0 or selected_name_index >= len(names):
            print("Invalid choice. Please select a number between 1 and", len(names))
        else:
            break
    except ValueError:
        print("Please enter a valid number.")

selected_name = names[selected_name_index]

```
