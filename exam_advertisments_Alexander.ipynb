```python
import pandas as pd


# https://www.kaggle.com/datasets/kavin1125/advertisment
# Load the CSV from Google Drive
url = 'https://drive.google.com/uc?id=1WZNsMHxcpuTcwHcxpPNNARNwCy2uJxhY'
data = pd.read_csv(url)


data.head()
```

```python
import numpy as np
import pandas as pd


# Combine relevant columns to create unique "arms"
data['arm'] = (
    data['Brand'] + "_" +
    data['Channel'] + "_" +
    data['Category'] + "_" +
    data['Spot_language']
)

# List of unique arms (ads)
arms = data['arm'].unique()
num_arms = len(arms)

# Initialize weights for each arm
weights = np.ones(num_arms)
gamma = 0.1  # Exploration rate

# Function to calculate probabilities for each arm
def calculate_probabilities(weights, gamma):
    total_weight = np.sum(weights)
    probabilities = (1 - gamma) * (weights / total_weight) + (gamma / num_arms)
    return probabilities

# Simulate rewards (replace this with real reward data if available)
# Here, we'll assume reward is a binary (e.g., 1 for clicked, 0 for not clicked)
np.random.seed(42)
rewards = np.random.choice([0, 1], size=len(data), p=[0.8, 0.2])  # Simulated data

# Mapping arm names to indices
arm_to_index = {arm: i for i, arm in enumerate(arms)}

# Exp3 Algorithm
T = len(data)  # Number of iterations
reward_history = []  # To store cumulative rewards

for t in range(T):
    # Step 1: Calculate probabilities
    probabilities = calculate_probabilities(weights, gamma)

    # Step 2: Select an arm based on probabilities
    current_arm = data['arm'].iloc[t]
    arm_index = arm_to_index[current_arm]
    selected_prob = probabilities[arm_index]

    # Step 3: Observe reward for the selected arm
    reward = rewards[t]  # Simulated reward (0 or 1)

    # Step 4: Update weights
    estimated_reward = reward / selected_prob
    weights[arm_index] *= np.exp((gamma * estimated_reward) / num_arms)

    # Track cumulative rewards
    reward_history.append(reward)

# Output the top-performing arms
top_arms = sorted(zip(arms, weights), key=lambda x: -x[1])[:5]
print("Top-performing arms (ads):")
for arm, weight in top_arms:
    print(f"Arm: {arm}, Weight: {weight}")
```
