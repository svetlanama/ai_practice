Протестуйте функцію на якомусь тексті.

```python

import nltk
from nltk import word_tokenize, pos_tag
from nltk.tree import Tree


def get_phrases(text, regex, tag_name):
    """
    Return list of phrases from text that matches regex.

    Params:
        text: str - original text
        regex: str - regular expression for grammar
        tag_name: str - tag name used in nltk tree

    Return:
        phrases: list[str] - list of phrases
    """
    # Tokenize text into words and apply part-of-speech tagging
    tokens = word_tokenize(text)
    tagged = pos_tag(tokens)

    # Create a RegexpParser with the given regex
    parser = nltk.RegexpParser(regex)

    # Parse the tagged text to create a chunk tree
    tree = parser.parse(tagged)

    # Extract phrases labeled with the specified tag_name
    phrases = []
    for subtree in tree:
        if isinstance(subtree, Tree) and subtree.label() == tag_name:
            phrase = " ".join(word for word, tag in subtree.leaves())
            phrases.append(phrase)
    return phrases



# Nount + Adjective definition
regex = r"""
    NP: {<JJ><NN>}
"""

phrases = get_phrases("A graceful deer leaps across the tranquil meadow under a golden sky..", regex, "NP")
print("Phrases:", phrases)

```
